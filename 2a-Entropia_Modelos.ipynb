# Calculo de la entropía de transferencia simbólica para los experimentos de de los modelos de conducción de calor

## Importación de bibliotecas

```python
from IPython import display
import ipywidgets
import numpy as np
import modulo_ets as mets
```

## Lectura de archivos

```python
ruta = "datos/experimentos_"
modelo = "fourier"
variable_x = "a"
variable_y = "b"
lista_x = np.loadtxt(ruta + modelo + "/" + "fluctuacion_" + variable_x + ".csv", delimiter = ",")[:, :-1]
lista_y = np.loadtxt(ruta + modelo + "/" + "fluctuacion_" + variable_y + ".csv", delimiter = ",")[:, :-1]
nlista = lista_x.shape[0]
```

```python
lista_x.shape
```

## Shuffle

```python
s_shuffle = 1
n_shuffle = int(lista_x.shape[1] / s_shuffle)
o_shuffle = np.arange(n_shuffle)
np.random.shuffle(o_shuffle)
o_shuffle
```

```python
lista_y = lista_y.reshape((144, n_shuffle, s_shuffle))[:, o_shuffle].reshape((144, n_shuffle * s_shuffle))
```

## Inicialización

```python
tv_ets = 3
nt_ets = 72
ets_xy = np.empty((nlista, nt_ets + 1))
ets_yx = np.empty((nlista, nt_ets + 1))
```

## Calculo de la curva de ETS

```python
# Barra de progreso
progreso = ipywidgets.FloatProgress(
    value = 0,
    min = 0,
    max = nlista, 
    description = 'Progreso:')
display.display(progreso)

# Proceso principal
progreso.value = 0
for e in range(nlista):
    ets_xy[e], ets_yx[e] = mets.causa(lista_x[e], lista_y[e], nt_ets, tv_ets)
    progreso.value = e + 1
```

## Escritura de archivos

```python
ruta = "datos/entropia_modelos/"
np.savetxt(ruta + modelo + "_" + variable_x + variable_y + "_null.csv", ets_xy[:, 1 :], delimiter = ",")
np.savetxt(ruta + modelo + "_" + variable_y + variable_x + "_null.csv", ets_yx[:, 1 :], delimiter = ",")
```
